```python
import qiskit
print(qiskit.__version__)
```

```python
from qiskit import QuantumCircuit
```

```python
qc = QuantumCircuit(2)
qc.draw("mpl")

qc.cx(0,1) # it works like takes 2 inputs (control, target) => (0,1) -> (1) ,    (1,0) -> 1
qc.x(0)
qc.z(0)
qc.h(1)
qc.cz(0,1)
qc.draw("mpl")
```

```python
from qiskit import quantum_info as qi
import numpy as np

c1 = qi.Statevector([1,0,1,0]/np.sqrt(2))
c1.draw("latex")
```

```python
c2 = QuantumCircuit(2)
quantum_state = qi.Statevector([1,0,0,1]/np.sqrt(2))

c2.initialize(quantum_state, [0,1])
c2.cx(0,1)
c2.z(1)
c2.draw("mpl")
```

```python
c3 = QuantumCircuit(2,2) # 2 quantum bits, and 2 classical bits for storing results of qubits.
c3.h(0)
c3.h(1)
c3.cx(0,1)
c3.h(0)
c3.h(1)
c3.measure(1,1) # measure qubit 1 and store it results in classical qubit 1.
c3.measure(0,0) # measure qubit 0 and store its result in classical qubit 0.


c3.draw('mpl')
```

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1)
qc.h(0)
qc.measure_all()
```

```python
from qiskit_aer import AerSimulator

simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()
print(counts)
```

```python
from qiskit.visualization import plot_distribution
plot_distribution(counts)
```
